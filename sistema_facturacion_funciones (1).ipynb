## 📦 Los datos (ya vienen listos)
Un diccionario para los productos y otro para los clientes.

````python
catalogo = {
    "C-010": {"nombre": "Cuaderno cuadriculado 100 hojas", "precio": 7500,  "iva": 0.19},
    "L-020": {"nombre": "Lápiz negro N.2",                 "precio": 1200,  "iva": 0.19},
    "B-030": {"nombre": "Libro escolar de matemáticas",     "precio": 45000, "iva": 0.00},
}

clientes = {
    "900123": {"nombre": "Colegio San Jose",     "tipo": "colegio"},
    "100987": {"nombre": "Laura Restrepo",       "tipo": "detal"},
    "900456": {"nombre": "Oficina Contadores SAS","tipo": "oficina"},
}

print("Productos:", len(catalogo), "· Clientes:", len(clientes))
````

## 🎁 Cómo se lee una función

```
def  nombre_de_la_funcion(parametros):   ← qué recibe
     """Qué hace"""                     ← documentación
     ...cuerpo...                        ← qué hace
     return resultado                    ← qué devuelve (opcional)
```
Una función que solo **muestra** información no necesita `return`.

## 📖 Misión 1 · `ver_catalogo()`

Convierte el diccionario del catálogo en una función que lo muestre.
No recibe nada, no devuelve nada: solo imprime.

````python
def ver_catalogo():
    """Muestra todos los productos del catálogo."""
    print("CATÁLOGO DE PRODUCTOS")
    for codigo, detalles in catalogo.items():
        print(f"Código: {codigo}, Nombre: {detalles['nombre']}, Precio: ${detalles['precio']:.2f}")


ver_catalogo()
````

## ➕ Misión 2 · `agregar_producto()`

Recibe código, nombre y precio, y guarda el producto en el catálogo.
Recuerda: `catalogo[codigo] = {...}` agrega una llave nueva al diccionario.

````python
def agregar_producto(codigo, nombre, precio):
    """Agrega un producto nuevo al catálogo."""
    catalogo[codigo] = {"nombre": nombre, "precio": precio, "iva": 0.19}
    print(f"Producto '{nombre}' (Código: {codigo}) agregado al catálogo.")

agregar_producto("T-040", "Memoria USB 32 GB", 38000)
ver_catalogo()
````

## 👥 Misión 3 · `ver_clientes()`

La misma idea de la misión 1, pero con el diccionario de clientes.

````python
def ver_clientes():
    """Muestra todos los clientes registrados."""
    print("\nLISTA DE CLIENTES")
    if not clientes:
        print("No hay clientes registrados.")
        return
    for documento, detalles in clientes.items():
        print(f"Documento: {documento}, Nombre: {detalles['nombre']}, Tipo: {detalles['tipo']}")

ver_clientes()
````

````python
def eliminar_cliente(documento):
    """Elimina un cliente del registro por su documento."""
    if documento in clientes:
        del clientes[documento]
        print(f"Cliente con documento {documento} eliminado correctamente.")
    else:
        print(f"Error: Cliente con documento {documento} no encontrado.")

eliminar_cliente("100987")
ver_clientes()
eliminar_cliente("999999") # Prueba de eliminación de cliente inexistente
ver_clientes()
````

## 🔍 Misión 5 · `consultar()`

Recibe un dato y lo busca primero en productos y luego en clientes.
Si no está en ninguno, avisa que no lo encontró.

````python
def consultar(dato):
    """Busca el dato en productos o en clientes."""
    if dato in catalogo:
        print("Producto:", catalogo[dato]["nombre"], "· $", catalogo[dato]["precio"])
    elif dato in clientes:
        print("Cliente:", clientes[dato]["nombre"])
    else:
        print("No encontré", dato)

consultar("C-010")
consultar("900123")
consultar("XXX")
````

## 🖥️ Misión 6 · `menu()` — la integradora

Aquí se juntan todas tus funciones. El menú ya no hace el trabajo: solo **dibuja las opciones**, pregunta y **llama** a la función correspondiente.

Así se verá en pantalla:
```
================================
   EL LÁPIZ FELIZ · CAJA
================================
   1. Ver catálogo
   2. Agregar producto
   3. Ver clientes
   4. Consultar
   5. Salir
```


````python
def menu():
    """Menú de la caja: solo coordina, no calcula."""
    opcion = ""
    while opcion != "5":
        print("=" * 32)
        print("   EL LÁPIZ FELIZ · CAJA")
        print("=" * 32)
        print("   1. Ver catálogo")
        print("   2. Agregar producto")
        print("   3. Ver clientes")
        print("   4. Consultar")
        print("   5. Salir")
        opcion = input("   Opción: ")

        # ✏️ TU CÓDIGO AQUÍ: un if/elif por cada opción, llamando a tus funciones
        if opcion == "1":
            ver_catalogo()
        elif opcion == "2":
            codigo = input("Código: ")
            nombre = input("Nombre: ")
            precio = float(input("Precio: "))
            agregar_producto(codigo, nombre, precio)
        elif opcion == "3":
            ver_clientes()
        elif opcion == "4":
            consultar(input("Código o documento: "))
        elif opcion == "5":
            print("   Hasta luego 👋")
        else:
            print("   Opción no válida")


menu()






menu()
````
